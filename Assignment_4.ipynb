```python
import pandas as pd
import numpy as np

# 1. Load dataset
df = pd.read_csv("winequality-red.csv", sep=";")
# Check missing values
print("Missing values before handling:")
print(df.isnull().sum())

# Handle missing values
df = df.fillna(df.mean(numeric_only=True))

print("Dataset Shape:", df.shape)
print("\nMissing values after handling:")
print(df.isnull().sum())

# Display the dataset
df.head()
```

```python
# 2. Extract vectors
alcohol_vector = df["alcohol"].to_numpy()
citric_acid_vector = df["citric acid"].to_numpy()

print("\nAlcohol Vector:")
print(alcohol_vector)

print("\nCitric Acid Vector:")
print(citric_acid_vector)
```

```python
# 3. Calculate covariance matrix
X = df[["alcohol", "density"]].to_numpy()

cov_matrix = np.cov(X.T)

print("\nCovariance Matrix:")
print(cov_matrix)
```

```python
# 4. Eigen decomposition
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

idx = np.argsort(eigenvalues)[::-1]

eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]
# Display eigenvalues
print("Eigenvalues:")
print(eigenvalues)

# Display eigenvectors
print("\nEigenvectors:")
print(eigenvectors)

# Sort eigenvalues in descending order
sorted_indices = np.argsort(eigenvalues)[::-1]

# Arrange eigenvalues from largest to smallest
top_eigenvalues = eigenvalues[sorted_indices]

# Arrange corresponding eigenvectors
top_eigenvectors = eigenvectors[:, sorted_indices]

# Display top 2 eigenvalues
print("\nTop 2 Eigenvalues:")
print(top_eigenvalues)

# Display corresponding eigenvectors
print("\nCorresponding Eigenvectors:")
print(top_eigenvectors)
```

```python
# 5. Wine quality distribution
quality_counts = df["quality"].value_counts()

print("\nWine Quality Distribution:")
print(quality_counts)

most_common_quality = quality_counts.idxmax()

print("\nMost Common Wine Quality:", most_common_quality)

# Print interpretation
print("\nInterpretation:")
print(f"Wine quality score {most_common_quality} is the most common in the dataset, with {quality_counts[most_common_quality]} samples.")

print("Most wines have quality scores between 5 and 6.")
print("Very few wines have quality scores of 3 or 8.")
print("This indicates that the dataset mainly contains medium-quality wines, with fewer low-quality and high-quality wines.")
```
